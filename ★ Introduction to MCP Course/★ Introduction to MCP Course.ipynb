# UV: Executive Summary

* Fast Python package/project manager (Rust-based, by Astral).
* Replaces pip, venv, pyenv, poetry.
* 10-100x faster installs than pip.

## Common Commands

* `uv init` → new project
* `uv add <package>` → install dependency
* `uv remove <package>` → uninstall dependency
* `uv run <file>.py` → run script
* `uv sync` → install exact lockfile versions
* `uv venv` → create virtual environment
* `uv python install <version>` → install Python version
* `uv lock` → update lockfile

```python
from mcp.client.stdio import StdioClientSession
from mcp.tools import Tool
import asyncio

# Connect to an MCP server using the stdio transport

# Initialize a connection to the MCP server
async def connect_to_mcp():
    """Establish a connection to an MCP server via stdio transport."""
    async with StdioClientSession() as session:
        # List available tools from the server
        tools_response = await session.list_tools()
        print(f"Available tools: {tools_response.tools}")
        return session

# Run the connection
asyncio.run(connect_to_mcp())
```
